# API de Clima

Proyecto de consumo de API listo para **Google Colab**.

**Pregunta:** ¿cómo obtener y visualizar datos del clima?

**Método:** requests a OpenWeatherMap (o similar)

**Stack:** requests, pandas, matplotlib

Ejemplo con API pública.

## 0. Setup

```python
import requests
import pandas as pd
import matplotlib.pyplot as plt
print("OK")
```

## 1. Ejemplo con API pública (Open-Meteo)

```python
# Open-Meteo no requiere API key
url = "https://api.open-meteo.com/v1/forecast"
params = {
    "latitude": -27.33,  # Curuzú Cuatiá
    "longitude": -58.05,
    "hourly": "temperature_2m",
    "forecast_days": 1
}
response = requests.get(url, params=params)
data = response.json()
print(data.keys())
```

## 2. Procesar datos

```python
horas = data["hourly"]["time"]
temps = data["hourly"]["temperature_2m"]

df = pd.DataFrame({"hora": : horas, "temperatura": : temps})
print(df.head())
```

## 3. Visualización

```python
plt.figure(figsize=(10, 4))
plt.plot(df["hora"], df["temperatura"], marker="o")
plt.title("Temperatura horaria")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 4. Conclusiones

- Open-Meteo es gratuita y no requiere API key.
- Ideal para proyectos de clima, agricultura o IoT.
- Se puede extender a más variables (lluvia, viento, etc.).

**Consejo:** para producción usá OpenWeatherMap o WeatherAPI con key.
